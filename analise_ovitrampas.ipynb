# Análise de imagens de ovitrampas

Este notebook localiza e lista imagens JPEG armazenadas em `data/images`.

```python
from pathlib import Path

pasta = Path("data/images")
imagens = sorted([
    *pasta.glob("*.jpg"),
    *pasta.glob("*.jpeg"),
    *pasta.glob("*.JPG"),
    *pasta.glob("*.JPEG"),
])

print(f"Pasta verificada: {pasta.resolve()}")
print(f"Imagens encontradas: {len(imagens)}")
for imagem in imagens[:10]:
    print(imagem.name)

if not pasta.exists():
    print("A pasta data/images ainda não existe. Crie-a na raiz do repositório e adicione as imagens.")
elif not imagens:
    print("Nenhum JPEG encontrado. Confira a extensão e se as imagens estão diretamente nesta pasta.")
```

```python
from PIL import Image
import matplotlib.pyplot as plt

amostra = imagens[:6]

fig, eixos = plt.subplots(2, 3, figsize=(15, 10))
for eixo, caminho in zip(eixos.flat, amostra):
    with Image.open(caminho) as img:
        eixo.imshow(img)
    eixo.set_title(caminho.name)
    eixo.axis("off")

for eixo in eixos.flat[len(amostra):]:
    eixo.axis("off")

plt.tight_layout()
plt.show()
```
